```python
import os
import pandas as pd
import matplotlib.pyplot as plt
```

```python
path = "../results"
results = os.listdir(path)
for file in results:
    df = pd.read_csv(f"{path}/{file}", header=None)
    plt.plot(df[1])

plt.xlabel("iterations")
plt.yscale("log")
plt.ylabel("best mse (log)")
plt.grid(True, which="both", alpha=0.3)
plt.legend(results)
```
